### Power of Two
[LeetCode:231](https://leetcode.com/problems/power-of-two/)

```python
class Solution:
    def isPowerOfTwoV2(self, n: int) -> bool:
        """
        Time: O(log n)
        Remarks:
            - Using recursion
        """
        if n == 1:
            return True
        if n < 1:
            return False

        return self.isPowerOfTwo(n / 2)

    def isPowerOfTwo(self, n: int) -> bool:
        """
        Time: O(1)
        Remarks:
            - Using bit manipulation
        """
        if n < 1:
            return False

        return n & (n - 1) == 0
```

### Number of 1 Bits
[LeetCode:191](https://leetcode.com/problems/number-of-1-bits/)

```python
class Solution:
    def hammingWeight(self, n: int) -> int:
        count = 0
        while n:
            count += 1
            n = n & (n - 1)

        return count
```

### Counting Bits
[LeetCode:338](https://leetcode.com/problems/counting-bits/)

```python
class Solution:
    def countBitsV1(self, n: int) -> list[int]:
        """
        Time: O(N × log(N))
        Space: O(N)
        Note:
            - A number N requires log2(N) bits for representation.
            - For each number, count set bits using Brian Kernighan’s algorithm.
        """
        result = []
        for number in range(n + 1):
            count = 0
            while number:
                count += 1
                number = number & (number - 1)  # clear lowest set bit
            result.append(count)

        return result

    def countBitsV2(self, n: int) -> list[int]:
        """
        Time: O(N)
        Space: O(N)
        Approach:
            - Use DP and bit properties.
            - answer[i] = answer[i // 2] + (i % 2).
        Property used:
            - Even numbers have the same bit count as i//2 (right shift).
            - Odd numbers have one more bit than i//2.
        """
        answer = [0] * (n + 1)

        for i in range(1, n + 1):
            if i & 1:  # Odd: one more bit than i//2
                answer[i] = answer[i // 2] + 1
            else:  # Even: same as i//2
                answer[i] = answer[i // 2]

        return answer


Solution().countBitsV2(3)
```

### Single number III
[LeetCode:260](https://leetcode.com/problems/single-number-iii/)

```python
class Solution:
    def singleNumber(self, nums: list[int]) -> list[int]:
        """
        Time: O(N)
        Space: O(1)
        Approach:
            - XOR cancels duplicates, leaving xor of two uniques.
            - Use rightmost set bit to divide numbers;
            - XOR within each group recovers the uniques.
        """
        xor = 0
        for n in nums:
            xor ^= n

        # Isolate the right most set-bit
        diff = xor & -xor

        a, b = 0, 0
        for n in nums:
            if n & diff:
                a ^= n
            else:
                b ^= n

        return [a, b]


Solution().singleNumber([2, 2, 6, 4, 7, 1, 1, 7])
```

### Add Binary
[LeetCode:67](https://leetcode.com/problems/add-binary/)

```python
class Solution:
    def addBinary(self, a: str, b: str) -> str:
        """
        Time: O(A+B)
        Space: O(1)
        Approach:
            - Convert strings to integers.
            - Use bitwise operations (XOR for sum, AND+shift for carry).
            - Repeat until no carry remains.
            - Convert result back to binary string.

        Why this works:
            - XOR computes the sum without carry.
            - AND finds positions where carry is needed, shifted left.
            - Repeating this process simulates manual binary addition.
        """
        a, b = int(a, 2), int(b, 2)

        while b:
            without_carry = a ^ b
            carry = (a & b) << 1
            a, b = without_carry, carry

        return bin(a)[2:]


Solution().addBinary("10010", "110")
```
